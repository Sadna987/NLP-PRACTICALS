# Q1: Prepare/Pre-process a text corpus to make it more usable for NLP tasks using tokenization, conversion to lowercase, removal of punctuation, filtration of stop words, stemming and lemmatization. 

```python
import nltk
import string
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('punkt_tab')
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
text="Natural Language Processing (NLP) is a field of AI. ## AI age"
tokens=word_tokenize(text)
print(tokens)
```

```python
tokens_lower=[word.lower() for word in tokens]
print(tokens_lower)
```

```python
tokens_wo_punkt=[word for word in tokens_lower if word not in string.punctuation]
print(tokens_wo_punkt)
```

```python
stopwords_list=stopwords.words('english')
tokens_wo_stopwords=[word for word in tokens_wo_punkt if word not in stopwords_list]
print(tokens_wo_stopwords)
```

```python
stemmer=PorterStemmer()
stemmed_words=[stemmer.stem(word) for word in tokens_wo_stopwords]
print(stemmed_words)
```

```python
lemmatizer=WordNetLemmatizer()
lemmatized_words=[lemmatizer.lemmatize(word) for word in tokens_wo_stopwords]
print(lemmatized_words)
```

```python
abbreviation_map = {
    "u": "you",
    "btw": "by the way",
    "idk": "I don't know",
    "gr8": "great"
}

def expand_abbreviations(text):
    words = text.split()
    expanded_words = [abbreviation_map.get(word, word) for word in words]
    return " ".join(expanded_words)

text = "u r gr8 btw"
print(expand_abbreviations(text))
```

# Q3: List the most common words (with their frequency) in a given text excluding stopwords. 

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter
import string

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab') 

def get_most_common_words(text, num_words=10):
    
    text = text.lower()

    # Tokenize the text
    words = word_tokenize(text)

    # Remove punctuation and non-alphabetic words
    words = [word for word in words if word.isalpha()]

    # Get English stopwords
    stop_words = set(stopwords.words('english'))

    # Remove stopwords
    filtered_words = [word for word in words if word not in stop_words]

    # Count word frequencies
    word_counts = Counter(filtered_words)

    # Return the most common words and their frequencies
    return word_counts.most_common(num_words)

# Example usage:
example_text = "This is an example sentence to demonstrate finding the most common words, excluding stopwords. This sentence has multiple words and some common words repeated."

print("Original Text:", example_text)
print("\nMost common words (excluding stopwords):\n", get_most_common_words(example_text, 5))

another_text = "Natural language processing is a field of artificial intelligence. It helps computers understand and process human language."
print("\nOriginal Text:", another_text)
print("\nMost common words (excluding stopwords):\n", get_most_common_words(another_text, 3))
```

# Q5: Build a simple statistical language model: estimate unigram and bigram probabilities with add one smoothing and compute the probability of given sentences.


```python

import nltk
from collections import Counter
import math

# Download tokenizer (run once)
nltk.download('punkt')

# Sample training corpus
corpus = [
    "I love machine learning",
    "I love deep learning",
    "machine learning is fun",
    "deep learning is powerful"
]

# Tokenization
tokens = []
for sentence in corpus:
    words = ["<s>"] + nltk.word_tokenize(sentence.lower()) + ["</s>"]
    tokens.extend(words)

# Vocabulary
vocab = set(tokens)
V = len(vocab)

# Unigram counts
unigram_counts = Counter(tokens)

# Bigram counts
bigrams = list(nltk.bigrams(tokens))
bigram_counts = Counter(bigrams)

# Total tokens
total_tokens = len(tokens)

# ===============================
#  Unigram Probability (Laplace)
# ===============================
def unigram_prob(word):
    return (unigram_counts[word] + 1) / (total_tokens + V)

# ===============================
#  Bigram Probability (Laplace)
# ===============================
def bigram_prob(w1, w2):
    return (bigram_counts[(w1, w2)] + 1) / (unigram_counts[w1] + V)

# ===============================
#  Sentence Probability
# ===============================
def sentence_probability(sentence):
    words = ["<s>"] + nltk.word_tokenize(sentence.lower()) + ["</s>"]
    
    prob = 1.0
    for i in range(len(words) - 1):
        prob *= bigram_prob(words[i], words[i+1])
    
    return prob

# ===============================
#  Test Sentences
# ===============================
test_sentences = [
    "I love learning",
    "deep learning is fun",
    "machine learning is powerful"
]

# ===============================
#  Output
# ===============================
print("Unigram Probabilities:")
for word in ["i", "learning", "machine"]:
    print(f"P({word}) = {unigram_prob(word):.4f}")

print("\nBigram Probabilities:")
print(f"P(love | i) = {bigram_prob('i','love'):.4f}")
print(f"P(learning | deep) = {bigram_prob('deep','learning'):.4f}")

print("\nSentence Probabilities:")
for sent in test_sentences:
    print(f"P('{sent}') = {sentence_probability(sent):.8f}")
```

# Q6: Perform POS tagging in a given text file. Extract all the nouns present in the text. Create and print a dictionary with frequency of parts of speech present in the document. 


```python

import nltk
from nltk import word_tokenize, pos_tag
from collections import Counter

import nltk

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')  # for newer NLTK versions
nltk.download('stopwords')
file_path = "input.txt"

with open(file_path, "r", encoding="utf-8") as f:
    text = f.read()

print(text)
tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)

pos_tags
nouns = [word for word, tag in pos_tags if tag.startswith('NN')]

print("Nouns in the text:")
print(nouns)
pos_counts = Counter(tag for word, tag in pos_tags)

print("POS Frequency Dictionary:")
print(pos_counts)
import pandas as pd

df_pos = pd.DataFrame(pos_counts.items(), columns=["POS Tag", "Frequency"])
df_pos.sort_values(by="Frequency", ascending=False)
```

# Q7: Identify and print the named entities using Name Entity Recognition (NER) for a collection of news headlines. 


```python

import spacy
!python -m spacy download en_core_web_sm
nlp = spacy.load("en_core_web_sm")
headlines = [
    "Apple launches new iPhone in California",
    "Elon Musk announces Tesla factory in India",
    "India defeats Australia in World Cup final",
    "Microsoft acquires AI startup in London",
    "NASA plans mission to Mars in 2027"
]
for i, headline in enumerate(headlines, start=1):
    doc = nlp(headline)
    print(f"\nHeadline {i}: {headline}")
    print("Named Entities:")
    for ent in doc.ents:
        print(f"  - {ent.text} → {ent.label_}")
import pandas as pd
entities = []
for headline in headlines:
    doc = nlp(headline)
    for ent in doc.ents:
        entities.append([headline, ent.text, ent.label_])
df = pd.DataFrame(entities, columns=["Headline", "Entity", "Label"])
df
```

# Q8. Classify movie reviews as positive or negative from the IMDB movie dataset of 5OK movie reviews. 


```python

import pandas as pd

df = pd.read_csv(r"C:\Users\ss724\Downloads\IMDB Dataset.csv")
df.head()
print(df.shape)
print(df['sentiment'].value_counts())
df['sentiment'] = df['sentiment'].map({'positive': 1, 'negative': 0})
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df['review'],
    df['sentiment'],
    test_size=0.2,
    random_state=42
)
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words='english', max_features=5000)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, y_train)
y_pred = model.predict(X_test_tfidf)
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
review = ["This movie was absolutely fantastic and very touching"]

review_tfidf = vectorizer.transform(review)
prediction = model.predict(review_tfidf)

print("Prediction:", "Positive" if prediction[0] == 1 else "Negative")
```
